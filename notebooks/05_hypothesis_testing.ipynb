```python
# Importing the relevant packages

import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency

# Connecting Google Drive to Colab

from google.colab import drive

drive.mount("/content/drive")

# Loading the QLFS Q2 2026 dataset from Google Drive

filename = "/content/drive/MyDrive/South_Africa_Labour_Market/QLFS202602.csv"

df = pd.read_csv(filename)

print("Dataset loaded successfully.")
print("Shape:", df.shape)

# Selecting the working-age population

working_age_df = df[
    (df["Q14AGE"] >= 15) &
    (df["Q14AGE"] <= 64)
].copy()


# Selecting employed and unemployed people only

test_df = working_age_df[
    working_age_df["Status"].isin([1, 2])
].copy()


# Creating readable labels

gender_labels = {
    1: "Male",
    2: "Female"
}

status_labels = {
    1: "Employed",
    2: "Unemployed"
}

test_df["gender_name"] = test_df["Q13GENDER"].map(gender_labels)
test_df["status_name"] = test_df["Status"].map(status_labels)


# Creating the contingency table

contingency_table = pd.crosstab(
    test_df["gender_name"],
    test_df["status_name"]
)

display(contingency_table)

# Checking the Status variable before hypothesis testing

print("Status data type:")
print(df["Status"].dtype)

print("\nFirst unique values:")
print(df["Status"].unique()[:10])

print("\nNumber of working-age observations:")
print(len(working_age_df))

# Converting the Status variable to numeric values

working_age_df["Status"] = pd.to_numeric(
    working_age_df["Status"],
    errors="coerce"
)

print("Status values after conversion:")
print(working_age_df["Status"].value_counts().head(10))

# Selecting employed and unemployed people only

test_df = working_age_df[
    working_age_df["Status"].isin([1, 2])
].copy()


# Creating readable labels

gender_labels = {
    1: "Male",
    2: "Female"
}

status_labels = {
    1: "Employed",
    2: "Unemployed"
}

test_df["gender_name"] = (
    test_df["Q13GENDER"]
    .map(gender_labels)
)

test_df["status_name"] = (
    test_df["Status"]
    .map(status_labels)
)


# Creating the contingency table

contingency_table = pd.crosstab(
    test_df["gender_name"],
    test_df["status_name"]
)

display(contingency_table)

# Performing a chi-square test of independence

chi2, p_value, dof, expected = chi2_contingency(
    contingency_table
)

print("Chi-square statistic:", round(chi2, 3))
print("P-value:", round(p_value, 5))
print("Degrees of freedom:", dof)

expected_table = pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
)

print("\nExpected frequencies:")
display(expected_table)

# Interpreting the chi-square test result

alpha = 0.05

if p_value < alpha:
    print("Reject H0.")
    print(
        "There is a statistically significant association "
        "between gender and employment status."
    )
else:
    print("Fail to reject H0.")
    print(
        "There is no statistically significant association "
        "between gender and employment status."
    )

    # Creating readable province labels

province_labels = {
    1: "Western Cape",
    2: "Eastern Cape",
    3: "Northern Cape",
    4: "Free State",
    5: "KwaZulu-Natal",
    6: "North West",
    7: "Gauteng",
    8: "Mpumalanga",
    9: "Limpopo"
}

test_df["province_name"] = (
    test_df["Province"]
    .map(province_labels)
)


# Creating the province and employment status contingency table

province_table = pd.crosstab(
    test_df["province_name"],
    test_df["status_name"]
)

display(province_table)

# Performing a chi-square test for province and employment status

chi2_province, p_province, dof_province, expected_province = (
    chi2_contingency(province_table)
)

print("Chi-square statistic:", round(chi2_province, 3))
print("P-value:", round(p_province, 5))
print("Degrees of freedom:", dof_province)

# Interpreting the province chi-square test result

alpha = 0.05

if p_province < alpha:
    print("Reject H0.")
    print(
        "There is a statistically significant association "
        "between province and employment status."
    )
else:
    print("Fail to reject H0.")
    print(
        "There is no statistically significant association "
        "between province and employment status."
    )
```
