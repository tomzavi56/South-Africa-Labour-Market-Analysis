```python
# Connecting Google Drive to Colab

from google.colab import drive

drive.mount("/content/drive")

# Loading the QLFS Q2 2026 dataset from Google Drive

import pandas as pd

filename = "/content/drive/MyDrive/South_Africa_Labour_Market/QLFS202602.csv"

df = pd.read_csv(filename)

print("Dataset loaded successfully.")
print("Shape:", df.shape)

display(df.head())

# Selecting the relevant variables for the analysis

analysis_df = df[
    [
        "Province",
        "Q13GENDER",
        "Q14AGE",
        "Q15POPULATION",
        "Age_grp1",
        "Education_status",
        "Status",
        "Lfs_Status",
        "Neet",
        "Weight"
    ]
].copy()

print("Selected dataset shape:", analysis_df.shape)

display(analysis_df.head())

# Replacing not-applicable values with missing values

not_applicable = "1.79769313486232e+308"

analysis_df[
    ["Status", "Lfs_Status", "Neet"]
] = analysis_df[
    ["Status", "Lfs_Status", "Neet"]
].replace(not_applicable, np.nan)


# Converting labour market variables to numeric values

analysis_df["Status"] = pd.to_numeric(
    analysis_df["Status"],
    errors="coerce"
)

analysis_df["Lfs_Status"] = pd.to_numeric(
    analysis_df["Lfs_Status"],
    errors="coerce"
)

analysis_df["Neet"] = pd.to_numeric(
    analysis_df["Neet"],
    errors="coerce"
)


# Checking the cleaned dataset

print("Data types:")
print(analysis_df.dtypes)

print("\nMissing values:")
print(analysis_df.isnull().sum())

display(analysis_df.head())

# Selecting the working-age population

working_age_df = analysis_df[
    (analysis_df["Q14AGE"] >= 15) &
    (analysis_df["Q14AGE"] <= 64)
].copy()

print("Original dataset:", analysis_df.shape)
print("Working-age dataset:", working_age_df.shape)

display(working_age_df.head())

# Creating readable labels for key variables

province_labels = {
    1: "Western Cape",
    2: "Eastern Cape",
    3: "Northern Cape",
    4: "Free State",
    5: "KwaZulu-Natal",
    6: "North West",
    7: "Gauteng",
    8: "Mpumalanga",
    9: "Limpopo"
}

gender_labels = {
    1: "Male",
    2: "Female"
}

status_labels = {
    1: "Employed",
    2: "Unemployed",
    3: "Discouraged jobseeker",
    4: "Other not economically active"
}

education_labels = {
    1: "No schooling",
    2: "Less than primary completed",
    3: "Primary completed",
    4: "Secondary not completed",
    5: "Secondary completed",
    6: "Tertiary",
    7: "Other"
}

working_age_df["province_name"] = working_age_df["Province"].map(province_labels)
working_age_df["gender_name"] = working_age_df["Q13GENDER"].map(gender_labels)
working_age_df["status_name"] = working_age_df["Status"].map(status_labels)
working_age_df["education_name"] = working_age_df["Education_status"].map(education_labels)

display(
    working_age_df[
        [
            "province_name",
            "gender_name",
            "Q14AGE",
            "education_name",
            "status_name",
            "Weight"
        ]
    ].head()
)

# Calculating weighted unemployment rates by province

labour_force_df = working_age_df[
    working_age_df["Status"].isin([1, 2])
].copy()

labour_force_df["weighted_unemployed"] = np.where(
    labour_force_df["Status"] == 2,
    labour_force_df["Weight"],
    0
)

provincial_unemployment = (
    labour_force_df
    .groupby("province_name")
    .agg(
        labour_force_weight=("Weight", "sum"),
        unemployed_weight=("weighted_unemployed", "sum")
    )
    .reset_index()
)

provincial_unemployment["unemployment_rate"] = (
    provincial_unemployment["unemployed_weight"]
    / provincial_unemployment["labour_force_weight"]
) * 100

provincial_unemployment = (
    provincial_unemployment
    .sort_values("unemployment_rate", ascending=False)
    .reset_index(drop=True)
)

display(provincial_unemployment)

# Visualising weighted unemployment rates by province

plt.figure(figsize=(10, 6))

plt.barh(
    provincial_unemployment["province_name"],
    provincial_unemployment["unemployment_rate"]
)

plt.xlabel("Unemployment Rate (%)")
plt.ylabel("Province")
plt.title("South African Unemployment Rate by Province, Q2 2026")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

# Calculating weighted unemployment rates by gender

gender_unemployment = (
    labour_force_df
    .groupby("gender_name")
    .agg(
        labour_force_weight=("Weight", "sum"),
        unemployed_weight=("weighted_unemployed", "sum")
    )
    .reset_index()
)

gender_unemployment["unemployment_rate"] = (
    gender_unemployment["unemployed_weight"]
    / gender_unemployment["labour_force_weight"]
) * 100

display(gender_unemployment)

# Visualising weighted unemployment rates by gender

plt.figure(figsize=(7, 5))

plt.bar(
    gender_unemployment["gender_name"],
    gender_unemployment["unemployment_rate"]
)

plt.title("South African Unemployment Rate by Gender, Q2 2026")
plt.xlabel("Gender")
plt.ylabel("Unemployment Rate (%)")
plt.tight_layout()
plt.show()

# Creating readable age-group labels

age_labels = {
    4: "15-19",
    5: "20-24",
    6: "25-29",
    7: "30-34",
    8: "35-39",
    9: "40-44",
    10: "45-49",
    11: "50-54",
    12: "55-59",
    13: "60-64"
}

labour_force_df["age_group_name"] = (
    labour_force_df["Age_grp1"]
    .map(age_labels)
)

# Calculating weighted unemployment rates by age group

age_unemployment = (
    labour_force_df
    .groupby("age_group_name")
    .agg(
        labour_force_weight=("Weight", "sum"),
        unemployed_weight=("weighted_unemployed", "sum")
    )
    .reset_index()
)

age_unemployment["unemployment_rate"] = (
    age_unemployment["unemployed_weight"]
    / age_unemployment["labour_force_weight"]
) * 100

display(age_unemployment)

# Visualising weighted unemployment rates by age group

plt.figure(figsize=(10, 5))

plt.bar(
    age_unemployment["age_group_name"],
    age_unemployment["unemployment_rate"]
)

plt.title("South African Unemployment Rate by Age Group, Q2 2026")
plt.xlabel("Age Group")
plt.ylabel("Unemployment Rate (%)")

plt.tight_layout()
plt.show()


# Calculating weighted unemployment rates by education level

education_unemployment = (
    labour_force_df
    .groupby("education_name")
    .agg(
        labour_force_weight=("Weight", "sum"),
        unemployed_weight=("weighted_unemployed", "sum")
    )
    .reset_index()
)

education_unemployment["unemployment_rate"] = (
    education_unemployment["unemployed_weight"]
    / education_unemployment["labour_force_weight"]
) * 100

education_unemployment = (
    education_unemployment
    .sort_values("unemployment_rate", ascending=False)
    .reset_index(drop=True)
)

display(education_unemployment)

# Visualising weighted unemployment rates by education level

plt.figure(figsize=(10, 6))

plt.barh(
    education_unemployment["education_name"],
    education_unemployment["unemployment_rate"]
)

plt.xlabel("Unemployment Rate (%)")
plt.ylabel("Education Level")
plt.title("South African Unemployment Rate by Education Level, Q2 2026")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

# Creating the youth population aged 15 to 34

youth_df = working_age_df[
    (working_age_df["Q14AGE"] >= 15) &
    (working_age_df["Q14AGE"] <= 34)
].copy()


# Calculating the weighted NEET rate

youth_df["weighted_neet"] = np.where(
    youth_df["Neet"] == 1,
    youth_df["Weight"],
    0
)

neet_rate = (
    youth_df["weighted_neet"].sum()
    / youth_df["Weight"].sum()
) * 100

print("Weighted NEET rate for youth aged 15-34:")
print(round(neet_rate, 2), "%")

```
